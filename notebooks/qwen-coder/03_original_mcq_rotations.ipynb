# Qwen Coder 7B: four MCQ rotations on Kaggle

Train a fresh **Qwen2.5-Coder-7B-Instruct** with 4-bit QLoRA and four balanced option rotations.
Each correct answer visits **a, b, c, d** once; the correct answer text is preserved.
Assistant targets are exactly `Answer:a`, `Answer:b`, `Answer:c`, or `Answer:d`.

**Before running:** create a private Kaggle dataset with `train.csv`, `validation.csv`,
and `TestData.csv` from this bundle. Attach it with **Add Input** (or **Add Data**).
Enable **Internet** and a **GPU** accelerator. This notebook uses one GPU even if two are assigned.
Test with the preflight/smoke cells, then use **Save Version → Save & Run All** for a fresh
background run. It needs no interactive uploads or clicks after execution starts.

The 1,718 training questions give up to 6,872 presentations before length filtering.
One epoch of this augmented set is **four original-question exposures**, exceeding the original
single-exposure assessment budget. Rotation reduces position bias; it cannot guarantee no overfitting.
Primary validation keeps its saved order; separate validation rotations measure position sensitivity.
Test answers are reserved for evaluation. Long groups are logged and excluded without truncation.

Results, adapter, logs and ZIP are written under `/kaggle/working` and are downloadable from the
saved version's output after completion. GPU quota and Kaggle session limits still apply.

```python
import os
# Set before importing torch/transformers: 4-bit training here deliberately uses one GPU.
os.environ["CUDA_VISIBLE_DEVICES"] = "0"
from pathlib import Path
import shutil

WORKING_DIR = Path("/kaggle/working")
INPUT_ROOT = Path("/kaggle/input")
if not WORKING_DIR.is_dir() or not INPUT_ROOT.is_dir():
    raise RuntimeError("Run this notebook on Kaggle with the CSV dataset attached.")
os.chdir(WORKING_DIR)
required_files = ["train.csv", "validation.csv", "TestData.csv"]
DATA_DIR = None  # Optional override, e.g. Path('/kaggle/input/my-mcq-dataset')
if DATA_DIR is None:
    candidates = sorted({p.parent for p in INPUT_ROOT.rglob("train.csv")
                         if all((p.parent / name).is_file() for name in required_files)})
    if len(candidates) != 1:
        raise RuntimeError(f"Expected one attached folder with all three CSVs; found {candidates}. Attach the dataset or set DATA_DIR.")
    DATA_DIR = candidates[0]
DATA_DIR = Path(DATA_DIR)
for name in required_files:
    source = DATA_DIR / name
    if not source.is_file():
        raise FileNotFoundError(source)
    shutil.copyfile(source, WORKING_DIR / name)
print("Input dataset:", DATA_DIR)
print("Working directory:", WORKING_DIR)
```

Parsed-content hashes verify the final question, option, answer and split data. UTF-8 BOM and Windows/Linux line-ending changes are accepted; actual data edits are rejected. Raw hashes are retained for provenance.
Changing only the option order requires changing the answer letter, never the correct answer text.


```python
import hashlib
import json
from collections import Counter
from finetune_qwen_mcq import load_split, check_split_isolation, rotate_question

def csv_content_hash(path):
    import csv
    with Path(path).open(encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        fields = ["class", "Question", "a", "b", "c", "d", "Answer"]
        if "source_row" in (reader.fieldnames or []):
            fields += ["source_row", "split"]
        if not set(fields).issubset(reader.fieldnames or []):
            raise ValueError(f"Missing required CSV columns in {path}")
        rows = [{key: row[key].replace("\r\n", "\n").replace("\r", "\n") for key in fields} for row in reader]
    content = json.dumps(rows, ensure_ascii=False, sort_keys=True, separators=(",", ":"))
    return hashlib.sha256(content.encode("utf-8")).hexdigest()

EXPECTED_HASHES = {'train.csv': 'c2f759271d8a7d4f29105bfae4b523c31b73101a6042be1db5b5c3ddebc8a58f', 'validation.csv': 'c7c028971d4ba185113eefafbb74260db4e5484542c04fec96334e1b18c77e20', 'TestData.csv': '5e10ec19d61624ee3554814337cf961fe499f4a666881c866c172fa1e51454a2'}
EXPECTED_CONTENT_HASHES = {'train.csv': 'ecc42e66eca12a63934f8929ba82b42190863f877cc8ca082c4dbeae1af468d8', 'validation.csv': '72dce6b6a0cf6df9e16abc969d7b56919dec6255ccf2d324bc84362bfc94d0ea', 'TestData.csv': '37851f92ed922296427ea2a18185a0c562335536fda2256f719a3804f0ba8b53'}
input_validation = {}
for name, expected in EXPECTED_HASHES.items():
    path = DATA_DIR / name
    actual_hash = hashlib.sha256(path.read_bytes()).hexdigest()
    content_hash = csv_content_hash(path)
    input_validation[name] = {"sha256": actual_hash, "content_sha256": content_hash,
                              "bytes_match": actual_hash == expected,
                              "content_matches": content_hash == EXPECTED_CONTENT_HASHES[name]}
    if not input_validation[name]["content_matches"]:
        raise ValueError(f"{name}: parsed question/option/answer content differs from the final dataset. This is more than an encoding or line-ending change; check that the complete CSV was copied.")
    if actual_hash != expected:
        print(f"{name}: file bytes changed, but parsed CSV contents match; accepted.")
    else:
        print(f"{name}: verified.")
if "EXPERIMENT_DIR" in globals():
    (EXPERIMENT_DIR / "input_validation.json").write_text(json.dumps(input_validation, indent=2), encoding="utf-8")

train_rows = load_split(DATA_DIR / "train.csv", "train")
validation_rows = load_split(DATA_DIR / "validation.csv", "validation")
check_split_isolation(train_rows, validation_rows)
print("Training questions:", len(train_rows), "-> presentations before length checks:", 4 * len(train_rows))
print("Validation questions:", len(validation_rows))
print("Four rotations of source", train_rows[0]["source_row"])
for row, mapping in rotate_question(train_rows[0], SEED):
    assert row[row["Answer"]] == train_rows[0][train_rows[0]["Answer"]]
    print("rotation", mapping["rotation"], "correct letter", row["Answer"], "new -> original", mapping["new_to_original"])
```

```python
import subprocess

def run_experiment(folder, *flags):
    command = [sys.executable, "-u", "finetune_qwen_mcq.py",
               "--train", str(DATA_DIR / "train.csv"), "--validation", str(DATA_DIR / "validation.csv"), "--test", str(DATA_DIR / "TestData.csv"),
               "--output", str(folder), "--max-length", str(MAX_LENGTH), "--seed", str(SEED), *flags]
    if SKIP_BASE_EVALUATION:
        command.append("--skip-base-evaluation")
    log_path = folder.parent / (folder.name + "-console.log")
    with log_path.open("w", encoding="utf-8") as log:
        with subprocess.Popen(command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                              text=True, bufsize=1) as process:
            for line in process.stdout:
                print(line, end="", flush=True)
                log.write(line)
                log.flush()
            result = process.wait()
    if result:
        raise RuntimeError(f"Run failed ({result}). Inspect {log_path}; partial results remain downloadable.")
```

## Results and position sensitivity

The main score uses the original test order. Rotation consistency is measured on held-out validation questions; a model can be consistently wrong, so inspect accuracy too.


```python
from IPython.display import Markdown, display

comparison = json.loads((RUN_DIR / "comparison.json").read_text())
metadata = json.loads((RUN_DIR / "metadata.json").read_text())
base, tuned = comparison["base"], comparison["finetuned"]
lines = ["| Metric | Base | Fine-tuned |", "|---|---:|---:|"]
for title, key in [("Original test accuracy", "test"), ("Fixed validation accuracy", "validation")]:
    a = f"{base[key]['accuracy']:.2%}" if base is not None else "Skipped"
    lines.append(f"| {title} | {a} | {tuned[key]['accuracy']:.2%} |")
for title, key in [("Consistent answer text across four orders", "consistent_answer_text_rate"),
                   ("Correct in all four orders", "all_four_correct_rate")]:
    a = f"{base['rotation'][key]:.2%}" if base is not None else "Skipped"
    lines.append(f"| {title} | {a} | {tuned['rotation'][key]:.2%} |")
for letter in "abcd":
    a = f"{base['rotation']['accuracy_by_correct_position'][letter]['accuracy']:.2%}" if base is not None else "Skipped"
    b = tuned["rotation"]["accuracy_by_correct_position"][letter]["accuracy"]
    lines.append(f"| Validation accuracy when correct answer is {letter} | {a} | {b:.2%} |")
display(Markdown("\n".join(lines)))
print("Baseline evaluation:", comparison.get("baseline_evaluation", "measured_this_run"))
print("Actual training presentations:", metadata["training_presentations"])
print("Four-exposure augmented experiment; not the original one-exposure assessment budget.")
print("Saved adapter reloaded:", comparison["saved_adapter_reload_verified"])
```

```python
import matplotlib.pyplot as plt

history = json.loads((RUN_DIR / "training_log.json").read_text())
train_points = [(r["step"], r["loss"]) for r in history if "loss" in r]
val_points = [(r["step"], r["eval_loss"]) for r in history if "eval_loss" in r]
fig, ax = plt.subplots(figsize=(8, 4))
if train_points:
    ax.plot(*zip(*train_points), label="Training loss")
if val_points:
    ax.scatter(*zip(*val_points), label="Fixed validation loss")
ax.set(xlabel="Optimizer step", ylabel="Assistant-token cross-entropy",
       title="One epoch over four rotations per question")
ax.legend()
fig.tight_layout()
fig.savefig(RUN_DIR / "loss.png", dpi=160)
plt.show()
```
